## Atividade 1
Escreva os métodos de ponto fixo dentro do arquivo $utils/algoritmos.py$.

````python
import numpy as np
from scipy.optimize import approx_fprime

def pontofixo(a, g, TOL=1e-8):
    x = g(a)
    while abs(x - a) > TOL:
        a = x
        x = g(a)
    return x

def newton_raphson(a, f, TOL=1e-8, df=None):
    if df is None:
        def dfn(x):
            return approx_fprime(np.array([x]), lambda v: f(v[0]))[0]
    else:
        dfn = df
    g = lambda x: x - f(x) / dfn(x)
    return pontofixo(a, g, TOL)

def secante(a, b, f, TOL=1e-8):
    g = lambda a, b: (a * f(b) - b * f(a)) / (f(b) - f(a))
    x = g(a, b)
    while abs(x - b) > TOL:
        a, b = b, x
        x = g(a, b)
    return x
````

Resolver a equação $e^x = x + 2$ é equivalente a calcular os pontos fixos da função  

$$
g(x) = e^x - 2
$$

Use os métodos do ponto fixo  

$$
x^{(n+1)} = g(x^{(n)})
$$

com $x^{(0)} = -1.8$ para obter uma aproximação de uma das soluções da equação dada com 8 dígitos significativos.  

**Resposta:**  
$
x \approx -1.8414057
$

````python
import numpy as np

import sys
import os

# Sobe um nível para a raiz do projeto e adiciona ao caminho do sistema
sys.path.append(os.path.abspath(os.path.join('..')))

from utils.algoritmos import (
    pontofixo,
    newton_raphson,
    secante,
)

f1 = lambda x: np.e**x - x - 2
g1 = lambda x: np.e**x - 2


def main():
    r = pontofixo(-1.8, g1)
    print(f"raiz ponto fixo = {r}")
    r = newton_raphson(-1.8, f1, df=lambda x: np.e**x - 1)
    print(f"raiz newton-raphson df = {r}")
    r = newton_raphson(-1.8, f1)
    print(f"raiz newton-raphson = {r}")
    r = secante(-1.8, -1.7, f1)
    print(f"raiz secante = {r}")


if __name__ == "__main__":
    main()
````

## Atividade 2
Encontre a raiz positiva da função  

$$
f(x) = \cos(x) - x^2
$$  

pelos métodos do ponto fixo, inicializando-o com $x^{(0)} = 1$.  

Realize a iteração até obter estabilidade no **quinto dígito significativo**.  

**Resposta:**

$$
x \approx 0.82413 
$$

Processo iterativo:  

$$
x^{(n+1)} = x^{(n)} + \frac{\cos(x) - x^2}{\sin(x) + 2x}
$$

````python
import sys
import os

# Sobe um nível para a raiz do projeto e adiciona ao caminho do sistema
sys.path.append(os.path.abspath(os.path.join('..')))

from utils.algoritmos import (
    pontofixo,
    newton_raphson,
    secante,
)

f2 = lambda x: np.cos(x) - x**2
g2 = lambda x: np.sqrt(np.cos(x))
df2 = lambda x: -np.sin(x) - 2*x

def main():
    r = pontofixo(1.0, g2, TOL=1e-5)
    print(f"raiz ponto fixo = {r:.5}")
    r = newton_raphson(1.0, f2, TOL=1e-5, df=df2)
    print(f"raiz newton-raphson df = {r:.5}")
    r = newton_raphson(1.0, f2, TOL=1e-5)
    print(f"raiz newton-raphson = {r:.5}")
    r = secante(1.0, 1.1, f2, TOL=1e-5)
    print(f"raiz secante = {r:.5}")


if __name__ == "__main__":
    main()
````

Aplique os métodos do ponto fixo para resolver a equação:

e^{-x^2} = 2x
$$

**Resposta:**

$$
x \approx 0.4193648
$$

````python
import numpy as np

import sys
import os

# Sobe um nível para a raiz do projeto e adiciona ao caminho do sistema
sys.path.append(os.path.abspath(os.path.join('..')))

from utils.algoritmos import (
    pontofixo,
    newton_raphson,
    secante,
)

f3 = lambda x: np.e**(-x**2) - 2*x
g3 = lambda x: np.e**(-x**2)/2
df3 = lambda x: (-2)*np.e**(-x**2) - 2

def main():
    r = pontofixo(1.0, g3)
    print(f"raiz ponto fixo = {r}")
    r = newton_raphson(1.0, f3, df=df3)
    print(f"raiz newton-raphson df = {r}")
    r = newton_raphson(1.0, f3)
    print(f"raiz newton-raphson = {r}")
    r = secante(1.0, 1.1, f3)
    print(f"raiz secante = {r}")


if __name__ == "__main__":
    main()
````

## Atividade 4
Resolva os três últimos exercícios da semana anterior usando os métodos do ponto fixo.

````python
import numpy as np

import sys
import os

# Sobe um nível para a raiz do projeto e adiciona ao caminho do sistema
sys.path.append(os.path.abspath(os.path.join('..')))

from utils.algoritmos import newton_raphson

def f4(x, V, R):
    IR = 1e-12  # corrente de saturação (A)
    T = 300.0  # temperatura (K)
    k = 1.38064852e-23  # constante de Boltzmann (J/K)
    q = 1.60217662e-19  # carga do elétron (C)
    vt = k * T / q  # tensão térmica (V)
    return R * IR * (np.exp(x / vt) - 1) + x - V

def f5(x):
    d = 500
    fmax = 50
    return (x*(np.cosh(d/(2*x))-1)) - fmax

def f6(x):
    F = 1e3
    L = 100e-3
    R = 1e3
    T = 2 * np.pi * F * L / R
    A = np.atan(T)
    return np.sin(x - A) + np.sin(A) * np.exp(-x / T)

def main():
    VRs = [
        (30, 1e3, 0, 1),
        (3, 1e3, 0, 1),
        (3, 1e4, 0, 1),
        (0.3, 1e3, 0, 0.5),
        (-0.3, 1e3, -1, 0),
        (-30, 1e3, -40, 0),
        (-30, 1e4, -40, 0),
    ]
    for V, R, a, b in VRs:
        try:
            f4_vrs = lambda x: f4(x, V, R)
            r = newton_raphson(b, f4_vrs)
            print(f"V={V} V, R={R/1e3:.0f}kΩ --> vd = {r:.3f} V")
        except ValueError as error:
            print(f"V={V} V, R={R/1e3:.0f}kΩ --> {error}")

    r = newton_raphson(550, f5)
    print(f"raiz = {r:.4}")
    
    a = 212 * np.pi / 180
    r = newton_raphson(a, f6)
    r_deg = r * 180 / np.pi
    print(f"raiz = {r_deg:.4}")


if __name__ == "__main__":
    main()
````
